```python
# --- [CELL 1]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 2}
# Reading dataset

df=pd.read_csv('data/data.csv')
```

```python
# --- [CELL 4]: ---
# cell_state: edited
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 5}
# === BEFORE (original) ===
# # Define X and y
# 
# #df.columns
# y=df['price']
# X=df[['bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot',
#        'floors', 'waterfront', 'view', 'condition', 'sqft_above',
#        'sqft_basement', 'yr_built', 'yr_renovated', 'street', 'city',
#        'statezip', 'country']]

# === AFTER (edited) ===
y = df['price']
X = df[['bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot',
        'floors', 'waterfront', 'view', 'condition', 'sqft_above',
        'sqft_basement', 'yr_built', 'yr_renovated', 'city']]
```

```python
# --- [CELL 5]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 6}
# Splitting the dataset

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.8,random_state=2529)
```

```python
# --- [CELL 6]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 7}
# Fitting the model
model.fit(X_train,y_train)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score

assert hasattr(model, "coef_"), "Model is not fitted"

# Shared contract for both fixes:
# all features used by the model must be numeric
assert len(X_train.select_dtypes(include=["object"]).columns) == 0, \
    f"Non-numeric columns remain in X_train: {X_train.select_dtypes(include=['object']).columns.tolist()}"
assert len(X_test.select_dtypes(include=["object"]).columns) == 0, \
    f"Non-numeric columns remain in X_test: {X_test.select_dtypes(include=['object']).columns.tolist()}"

# Predict successfully on both splits
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

assert np.isfinite(y_train_pred).all() and np.isfinite(y_test_pred).all(), \
    "Predictions contain non-finite values"
assert len(y_train_pred) == len(y_train), "Training prediction length mismatch"
assert len(y_test_pred) == len(y_test), "Test prediction length mismatch"

train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
train_r2 = r2_score(y_train, y_train_pred)
test_r2 = r2_score(y_test, y_test_pred)

assert np.isfinite(train_rmse) and np.isfinite(test_rmse), "RMSE is non-finite"
assert np.isfinite(train_r2) and np.isfinite(test_r2), "R2 is non-finite"
```
