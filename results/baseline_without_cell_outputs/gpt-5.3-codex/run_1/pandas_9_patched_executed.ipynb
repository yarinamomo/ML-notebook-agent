```python
# --- [CELL 1]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 2}
df = pd.read_csv("data/IMDb_All_Genres_etf_clean1.csv")
```

```python
# --- [CELL 2]: ---
# cell_state: edited
# execution_status: {'status': 'error', 'done': True, 'execution_count': 3}
# === BEFORE (original) ===
# clean_df = df[(df["Total Gross (millions)"]!="$0.00M") & (df["Total Gross (millions)"]!="Gross Unkown")].copy()
# clean_df = clean_df[clean_df['Censor']!="(Banned)"]

# === AFTER (edited) ===
# Normalize column names to handle slight schema differences
clean_df = df.copy()
clean_df.columns = clean_df.columns.str.strip()

# Resolve possible gross column name variants
gross_col = None
for c in ["Total Gross (millions)", "Total Gross", "total gross (millions)"]:
    if c in clean_df.columns:
        gross_col = c
        break

if gross_col is None:
    raise KeyError(f"Could not find expected gross column. Available columns: {list(clean_df.columns)}")

# Keep rows with known/non-zero gross values
clean_df = clean_df[(clean_df[gross_col] != "$0.00M") & (clean_df[gross_col] != "Gross Unkown")].copy()

# Apply censor filter if column exists
if "Censor" in clean_df.columns:
    clean_df = clean_df[clean_df["Censor"] != "(Banned)"]
```

```python
import pandas as pd

assert 'clean_df' in globals(), 'clean_df should be created by the previous cell.'
assert 'Total_Gross' in df.columns, 'Expected canonical gross column Total_Gross to exist.'

expected_clean_df = df[(df['Total_Gross'] != '$0.00M') & (df['Total_Gross'] != 'Gross Unkown')].copy()
expected_clean_df = expected_clean_df[expected_clean_df['Censor'] != '(Banned)'].copy()

pd.testing.assert_frame_equal(
    clean_df.reset_index(drop=True),
    expected_clean_df.reset_index(drop=True),
    check_dtype=False
)
```
