```python
# --- [CELL 0]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 1}
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# --- [CELL 1]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 2}
df = pd.read_csv("data/SalaryData_Test(1).csv")
```

```python
# --- [CELL 2]: ---
# cell_state: edited
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 3}
# === BEFORE (original) ===
# x = df.workclass.values.reshape(-1,1)
# y = df.Salary.values.reshape(-1,1)

# === AFTER (edited) ===
# Use numeric features for x (age, educationno, capitalgain, capitalloss, hoursperweek)
x = df[['age', 'educationno', 'capitalgain', 'capitalloss', 'hoursperweek']].values

# Convert salary to binary: 0 for <=50K, 1 for >50K
y = df['Salary'].apply(lambda x: 1 if x == ' >50K' else 0).values.reshape(-1, 1)
```

```python
# --- [CELL 3]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 4}
from sklearn.linear_model import LinearRegression

linear_reg = LinearRegression() 
linear_reg.fit(x,y)
```

```python
import numpy as np

# Regression fit should only proceed on numeric arrays after encoding.
assert np.issubdtype(x.dtype, np.number), f"x must be numeric, got {x.dtype}"
assert np.issubdtype(y.dtype, np.number), f"y must be numeric, got {y.dtype}"
assert x.ndim == 2 and y.ndim == 2, "x and y must be 2D column vectors"
assert x.shape[1] == 1 and y.shape[1] == 1, "x and y must each have one column"
assert hasattr(linear_reg, "coef_"), "linear_reg should be fitted and expose coef_"
assert linear_reg.coef_.size > 0, "fitted model should have non-empty coefficients"
```
