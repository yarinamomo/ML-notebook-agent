```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras.utils import to_categorical
from imblearn.over_sampling import SMOTE
from keras.optimizers import Adam

# Load the data
data = pd.read_csv("/kaggle/input/prudential-life-insurance-assessment/train.csv.zip")

missing_val_count_by_column = data.isnull().sum()/len(data)
a = missing_val_count_by_column[missing_val_count_by_column > 0.4].sort_values(ascending=False).index.tolist()
data.drop(columns = a, axis=1, inplace=True)
data.drop('Family_Hist_4', axis=1, inplace=True)

# Calculate the sum of all Medical_Keyword columns
medical_keywords = data.columns[data.columns.str.startswith('Medical_Keyword')]
data['Medical_Keyword_Sum'] = data[medical_keywords].sum(axis=1)

# Drop the original Medical_Keyword columns
data.drop(columns=medical_keywords, inplace=True)

# Separate features and target
X = data.drop('Response', axis=1)
y = data['Response']

# One-hot encoding for the 'Product_Info_2' column
X_encoded = pd.get_dummies(X)

# Split the data into train and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# Impute missing values with mean
imputer = SimpleImputer(strategy='mean')
X_train_imputed = imputer.fit_transform(X_train)
X_val_imputed = imputer.transform(X_val)

# Apply oversampling to handle class imbalance
smote = SMOTE(random_state=42)
X_train_oversampled, y_train_oversampled = smote.fit_resample(X_train_imputed, y_train)

# Feature scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_oversampled)
X_val_scaled = scaler.transform(X_val_imputed)

# One-hot encode the target variable
y_train_encoded = to_categorical(y_train_oversampled - 1)  # Subtract 1 to make the classes start from 0
y_val_encoded = to_categorical(y_val - 1)

# Create a Sequential model
model = Sequential()

# Add the input layer
model.add(Dense(256, input_dim=X_train_scaled.shape[1], activation='relu'))
model.add(Dropout(0.5))  # Drop about 50% of the nodes

# Add more hidden layers
model.add(Dense(128, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.2))  # Drop about 20% of the nodes
model.add(Dense(16, activation='relu'))

# Add the output layer
model.add(Dense(y_train_encoded.shape[1], activation='softmax'))

# Compile the model with a custom learning rate
opt = Adam(learning_rate=5e-4)
model.compile(loss='categorical_crossentropy', optimizer=opt, metrics=['accuracy'])

# Train the model for more epochs
model.fit(X_train_scaled, y_train_encoded, validation_data=(X_val_scaled, y_val_encoded), epochs=20, batch_size=32)
```
