```python
# --- [CELL 1]: ---
# cell_state: unchanged
# execution_status: {'status': 'not run'}
df = pd.read_csv("data/IMDb_All_Genres_etf_clean1.csv")
```

```python
# --- [CELL 2]: ---
# cell_state: edited
# execution_status: {'status': 'not run'}
# === BEFORE (original) ===
# clean_df = df[(df["Total Gross (millions)"]!="$0.00M") & (df["Total Gross (millions)"]!="Gross Unkown")].copy()
# clean_df = clean_df[clean_df['Censor']!="(Banned)"]

# === AFTER (edited) ===
# Check available columns
print("Available columns in the dataframe:")
print(df.columns.tolist())

# Try to find the gross column variants
gross_col = None
for col in df.columns:
    if 'gross' in col.lower() or 'Gross' in col:
        gross_col = col
        print(f"Found gross-related column: '{col}'")

if gross_col:
    clean_df = df[(df[gross_col]!="$0.00M") & (df[gross_col]!="Gross Unkown")].copy()
    
    # Check for Censor column
    censor_col = None
    for col in df.columns:
        if 'censor' in col.lower() or 'Censor' in col:
            censor_col = col
            print(f"Found censor-related column: '{col}'")
    
    if censor_col:
        clean_df = clean_df[clean_df[censor_col]!="(Banned)"]
else:
    print("No gross column found, creating copy of original df")
    clean_df = df.copy()
```

```python
import pandas as pd

assert 'clean_df' in globals(), 'clean_df should be created by the previous cell.'
assert 'Total_Gross' in df.columns, 'Expected canonical gross column Total_Gross to exist.'

expected_clean_df = df[(df['Total_Gross'] != '$0.00M') & (df['Total_Gross'] != 'Gross Unkown')].copy()
expected_clean_df = expected_clean_df[expected_clean_df['Censor'] != '(Banned)'].copy()

pd.testing.assert_frame_equal(
    clean_df.reset_index(drop=True),
    expected_clean_df.reset_index(drop=True),
    check_dtype=False
)
```
