```python
# --- [CELL 1]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 2}
df=pd.read_csv('data/hmeq.csv')
df.head()
```

```python
# --- [CELL 2]: ---
# cell_state: unchanged
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 3}
df.loc[df.BAD == 1, 'STATUS'] = 'DEFAULT'
df.loc[df.BAD == 0, 'STATUS'] = 'PAID'
```

```python
# --- [CELL 3]: ---
# cell_state: edited
# execution_status: {'status': 'ok', 'done': True, 'execution_count': 4}
# === BEFORE (original) ===
# # %%opts Histogram[width=700 height=400 tools=['hover'] xrotation=0]{+axiswise +framewise}
# 
# g = df.groupby('STATUS')
# 
# cols = ['LOAN',
#         'MORTDUE', 
#         'VALUE',
#         'YOJ',
#         'DEROG',
#         'DELINQ',
#         'CLAGE',
#         'NINQ',
#         'CLNO']
# dd={}
# 
# # Histograms
# for col in cols:
#     
#     freq, edges = np.histogram(df[col].values)
#     dd[col] = hv.Histogram((edges, freq), label='ALL Loans').redim.label(x=' ')
#     
#     freq, edges = np.histogram(g.get_group('PAID')[col].values, bins=edges)
#     dd[col] *= hv.Histogram((edges, freq), label='PAID Loans').redim.label(x=' ')
#     
#     freq, edges = np.histogram(g.get_group('DEFAULT')[col].values, bins=edges)
#     dd[col] *= hv.Histogram((edges, freq), label='DEFAULT Loans' ).redim.label(x=' ')   
#     
# var = [*dd]
# kdims=hv.Dimension(('var', 'Variable'), values=var)    
# hv.HoloMap(dd, kdims=kdims)

# === AFTER (edited) ===
g = df.groupby('STATUS')

cols = ['LOAN',
        'MORTDUE',
        'VALUE',
        'YOJ',
        'DEROG',
        'DELINQ',
        'CLAGE',
        'NINQ',
        'CLNO']
dd = {}

for col in cols:
    # Drop NaNs to avoid histogram range errors when a column/group is all-missing
    all_vals = df[col].dropna().values
    paid_vals = g.get_group('PAID')[col].dropna().values
    default_vals = g.get_group('DEFAULT')[col].dropna().values

    # If no valid values exist at all, skip this variable
    if all_vals.size == 0:
        continue

    freq, edges = np.histogram(all_vals)
    dd[col] = hv.Histogram((edges, freq), label='ALL Loans').redim.label(x=' ')

    if paid_vals.size > 0:
        freq, _ = np.histogram(paid_vals, bins=edges)
        dd[col] *= hv.Histogram((edges, freq), label='PAID Loans').redim.label(x=' ')

    if default_vals.size > 0:
        freq, _ = np.histogram(default_vals, bins=edges)
        dd[col] *= hv.Histogram((edges, freq), label='DEFAULT Loans').redim.label(x=' ')

var = [*dd]
kdims = hv.Dimension(('var', 'Variable'), values=var)
hv.HoloMap(dd, kdims=kdims)
```
